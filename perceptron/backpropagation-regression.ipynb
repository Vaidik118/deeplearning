```python
def initialize_parameters(layer_dims):
  
  np.random.seed(3)
  parameters = {}
  L = len(layer_dims)         

  for l in range(1, L):

    parameters['W' + str(l)] = np.ones((layer_dims[l-1], layer_dims[l]))*0.1
    parameters['b' + str(l)] = np.zeros((layer_dims[l], 1))
      

  return parameters
```

```python
initialize_parameters([2,2,1])
```

```python
def linear_forward(A_prev, W, b):
  
  Z = np.dot(W.T, A_prev) + b
  
  return Z
```

```python
# Forward Prop
def L_layer_forward(X, parameters):

  A = X
  L = len(parameters) // 2                  # number of layers in the neural network
  
  for l in range(1, L+1):
    A_prev = A 
    Wl = parameters['W' + str(l)]
    bl = parameters['b' + str(l)]
    #print("A"+str(l-1)+": ", A_prev)
    #print("W"+str(l)+": ", Wl)
    #print("b"+str(l)+": ", bl)
    #print("--"*20)

    A = linear_forward(A_prev, Wl, bl)
    #print("A"+str(l)+": ", A)
    #print("**"*20)
          
  return A,A_prev
```

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = df[['lpa']].values[0][0]

# Parameter initialization
parameters = initialize_parameters([2,2,1])

y_hat, A1 = L_layer_forward(X, parameters)
```

```python
def update_parameters(parameters, y, y_hat, A1, X):

    learning_rate = 0.001
    error = y - y_hat

    # Save old W2
    W2_old = parameters['W2'].copy()

    # -------------------------
    # Output Layer
    # -------------------------

    parameters['W2'][0][0] = (
        parameters['W2'][0][0]
        + learning_rate * 2 * error * A1[0][0]
    )

    parameters['W2'][1][0] = (
        parameters['W2'][1][0]
        + learning_rate * 2 * error * A1[1][0]
    )

    parameters['b2'][0][0] = (
        parameters['b2'][0][0]
        + learning_rate * 2 * error
    )

    # -------------------------
    # Hidden Layer
    # -------------------------

    parameters['W1'][0][0] = (
        parameters['W1'][0][0]
        + learning_rate * 2 * error * W2_old[0][0] * X[0][0]
    )

    parameters['W1'][0][1] = (
        parameters['W1'][0][1]
        + learning_rate * 2 * error * W2_old[0][0] * X[1][0]
    )

    parameters['b1'][0][0] = (
        parameters['b1'][0][0]
        + learning_rate * 2 * error * W2_old[0][0]
    )

    parameters['W1'][1][0] = (
        parameters['W1'][1][0]
        + learning_rate * 2 * error * W2_old[1][0] * X[0][0]
    )

    parameters['W1'][1][1] = (
        parameters['W1'][1][1]
        + learning_rate * 2 * error * W2_old[1][0] * X[1][0]
    )

    parameters['b1'][1][0] = (
        parameters['b1'][1][0]
        + learning_rate * 2 * error * W2_old[1][0]
    )
```

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = df[['lpa']].values[0][0]

# Parameter initialization
parameters = initialize_parameters([2,2,1])

y_hat,A1 = L_layer_forward(X,parameters)
y_hat = y_hat.item()

update_parameters(parameters,y,y_hat,A1,X)

parameters
```

```python
# epochs implementation

parameters = initialize_parameters([2, 2, 1])

epochs = 5

for i in range(epochs):

    Loss = []

    for j in range(df.shape[0]):

        X = df[['cgpa', 'profile_score']].values[j].reshape(2, 1)

        y = df[['lpa']].values[j][0]

        # Forward propagation
        y_hat, A1 = L_layer_forward(X, parameters)

        # Convert (1,1) array to scalar
        y_hat = y_hat.item()

        # Update parameters
        update_parameters(
            parameters,
            y,
            y_hat,
            A1,
            X
        )

        # Calculate loss
        Loss.append((y - y_hat) ** 2)

    print(
        'Epoch - ',
        i + 1,
        'Loss - ',
        np.array(Loss).mean()
    )

parameters
```
